## Une prévision proche peut-elle inverser le classement ?

Trois actions réalisent 4 %, 3 % et 2 %. Le modèle prévoit 2 %, 3 % et 4 %.
Calculez les erreurs au carré, puis comparez-les à celles de la prévision nulle.
Le calcul donne 8 contre 29 en pourcentages carrés. Le R² vaut donc 21/29, malgré un classement inversé.


```python
print("R²", examples["forecast_r2"])
print("Corrélation de classement", examples["forecast_rank"])
```

## Que changent deux entreprises disparues ?

Cinq placements de 100 dollars deviennent 120, 110, 90, 0 et 0 dollars.
Comparez la performance complète et celle calculée sur les trois premiers seulement.
La première vaut moins 36 %. La seconde vaut environ plus 6,67 %.


```python
print("Univers complet", examples["survival_complete"])
print("Trois survivantes", examples["survival_selected"])
```

## Pourquoi les rendements se composent-ils ?

Un prix passe de 100 à 102 dollars pendant la nuit, puis à 101 dollars à la clôture.
Le gain total vaut 1 %. Il faut multiplier les facteurs de croissance pour le retrouver.


```python
print("Nuit", examples["night_return"])
print("Séance", examples["session_return"])
print("Total", examples["daily_return"])
```
